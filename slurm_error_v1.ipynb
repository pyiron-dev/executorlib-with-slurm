## Submit Python Function to SLURM
This is just like using `sbatch` to submit shell scripts.

```python
def echo(i):
    return i
```

```python
with executorlib.SlurmClusterExecutor() as exe:
    f1 = exe.submit(
        echo,
        resource_dict={})
    result = f1.result()
    print(result)
```

```python
excep = f1.exception()
excep.output
```
